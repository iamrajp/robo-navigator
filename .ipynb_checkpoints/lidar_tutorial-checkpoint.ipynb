## Step 2 — Check the LiDAR is detected

Plug in the LiDAR via USB. It should appear as `/dev/ttyUSB0`.

```python
import os

port = '/dev/ttyUSB0'

if os.path.exists(port):
    print(f"✓ LiDAR detected at {port}")
else:
    print(f"✗ Not found at {port}")
    print("  → Check USB cable")
    print("  → Run: sudo usermod -aG dialout $USER  (then re-login)")
```

---
## Step 3 — Connect and read device info

```python
from rplidar import RPLidar

PORT     = '/dev/ttyUSB0'
BAUDRATE = 460800  # C1 specific — do not change to 115200

lidar = RPLidar(PORT, baudrate=BAUDRATE)

print("Device info:", lidar.get_info())
print("Health:     ", lidar.get_health())
```

---
## Step 4 — Start the motor and take ONE scan

A single scan = one full 360° sweep.  
Each point is `(quality, angle_degrees, distance_mm)`.

```python
import time

lidar.start_motor()
time.sleep(1)  # wait for motor to reach full speed

# grab just one scan
scan = next(lidar.iter_scans())

print(f"Points in this scan: {len(scan)}")
print("\nFirst 10 points (quality, angle°, distance mm):")
for point in scan[:10]:
    quality, angle, distance = point
    print(f"  Q={quality:3d}  angle={angle:6.1f}°  dist={distance:7.1f} mm")
```

---
## Step 5 — Understand the data

| Field | Meaning |
|-------|---------|
| `quality` | Signal strength (0 = invalid, higher = better) |
| `angle` | Direction in degrees (0–360, clockwise from front) |
| `distance` | Distance to object in **millimetres** |

Filter out bad points (quality=0 or distance=0):

```python
good_points = [(q, a, d) for q, a, d in scan if q > 0 and d > 0]

print(f"Total points : {len(scan)}")
print(f"Valid points : {len(good_points)}")

distances = [d for _, _, d in good_points]
print(f"Min distance : {min(distances):.0f} mm ({min(distances)/1000:.2f} m)")
print(f"Max distance : {max(distances):.0f} mm ({max(distances)/1000:.2f} m)")
```

---
## Step 6 — Visualise the scan (polar plot)

A polar plot shows exactly what the LiDAR "sees" around it.

```python
import math
import matplotlib.pyplot as plt

angles    = [math.radians(a) for _, a, _ in good_points]
distances = [d / 1000        for _, _, d in good_points]  # mm → metres

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'}, figsize=(7, 7))
ax.scatter(angles, distances, s=2, c='cyan')
ax.set_title('LiDAR 360° Scan', pad=15)
ax.set_theta_zero_location('N')   # 0° at top (North)
ax.set_theta_direction(-1)        # clockwise
plt.tight_layout()
plt.show()
```

---
## Step 7 — Live continuous scanning

Loop over scans in real time. Press the **Stop** button (■) in JupyterLab to interrupt.

```python
import math
import matplotlib.pyplot as plt
from IPython.display import clear_output

try:
    for i, scan in enumerate(lidar.iter_scans()):
        good = [(q, a, d) for q, a, d in scan if q > 0 and d > 0]
        if not good:
            continue

        angles    = [math.radians(a) for _, a, _ in good]
        distances = [d / 1000        for _, _, d in good]

        clear_output(wait=True)
        fig, ax = plt.subplots(subplot_kw={'projection': 'polar'}, figsize=(7, 7))
        ax.scatter(angles, distances, s=2, c='cyan')
        ax.set_title(f'Live Scan #{i+1}  ({len(good)} points)', pad=15)
        ax.set_theta_zero_location('N')
        ax.set_theta_direction(-1)
        plt.tight_layout()
        plt.show()

except KeyboardInterrupt:
    print("Stopped.")
```

---
## Step 8 — Always disconnect when done

**Always run this cell when finished** — leaving the motor spinning wears it out.

```python
lidar.stop()
lidar.stop_motor()
lidar.disconnect()
print("LiDAR disconnected safely.")
```

---
## Troubleshooting

| Problem | Fix |
|---------|-----|
| `Permission denied /dev/ttyUSB0` | `sudo usermod -aG dialout $USER` then re-login |
| No data / garbage data | Wrong baud rate — must be `460800` |
| Device not found | Check USB cable; `ls /dev/ttyUSB*` |
| Scan freezes | Call `lidar.stop()` then reconnect |
| Motor doesn't spin | Try a powered USB hub (Pi may not supply enough current) |
